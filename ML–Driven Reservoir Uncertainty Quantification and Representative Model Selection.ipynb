## 1. Import needed libraries / dependencies

```python
import lasio
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
import xgboost as xgb

import warnings
warnings.filterwarnings('ignore')

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
```

## 2. Load the data and clean

```python
las = lasio.read('https://kgsimages.blob.core.windows.net/web/web_1/WebDocs/WellLogs/kcc_logs_2025/1056890332.las')
df = las.df()
df.head()
```

```python
df.info()
```

```python
# rearrange the dataframe and make depth a distinct column.
# ALso correct spelling from DEPT to Depth

df = df.reset_index()
df= df.rename(columns={'DEPT':'DEPTH'})
df.head()
```

```python
# drop nulls in the column of interest (GR, NPHI, DPHI, ILD, CALM)
df = df.dropna(subset = ['GR', 'NPHI', 'DPHI', 'ILD', 'CALM'])
df.head()
```

```python
df.info()
```

## 3. Delve into advanced petrophysical processing

### 3a. Calculate volume of shale

```python
# Compute Shale volume, Vsh, using the simple linear method  (simple linear GR method - common in petrophysics)
gr_min  = df['GR'].quantile(0.05) #This represents the gamma ray value for clean sand reservoir
gr_max = df['GR'].quantile(0.95) # This represents the gamma ray value for a shale lithology
df['VSH'] = (df['GR'] - gr_min) / (gr_max - gr_min)
df['VSH'] = df['VSH'].clip(0, 1) # The clip metod is applied to ensure the computed Vshale falls between 0 and 1
```

```python
df.head()
```

### 3b. Calculate effective porosity by averaging density and neutron porosity after correcting for shale volume

```python
# 2.2 Effective porosity (average of density + neutron, shale-corrected)
df['PHI'] = ((df['DPHI'] + df['NPHI']) / 2) * (1 - df['VSH'])
df['PHI'] = df['PHI'].clip(0, 0.45)   # clip porosity value between 0 and 45% to make it realistic
df['PHI'].nunique()
```

### 3c. Feature Engineering for ML 

```python
# To select the features that will be used for the machine learning aspect

df['PHID_NPHID'] = df['DPHI'] - df['NPHI']   # crossover indicator
df['RHOB_proxy'] = 2.65 - (2.65 - 1.0) * df['PHI']  # a synthetic bulk density log

print(df[['DEPTH', 'PHI', 'VSH']].describe())
```

## 4. Rock Typing Using Unsupervised machine learning

```python
features_cluster = ['GR', 'DPHI', 'NPHI', 'ILD', 'PHI', 'VSH'] #These are the logs that are more critical in determinig the rock and fluid behavior of the interval being logged
X_cluster = df[features_cluster].copy()  # A new dataframe is created using these logs to ensure the original dataframe is not tampered with or modified

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cluster)  # The log values are standardised to ensure unifmroity since the sale of measurement of each log s different

# 5 clusters is decided to be sed for my K means model, and the algorithm is selected to run 10 times and oick the best model
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df['ELECTROFACIES'] = kmeans.fit_predict(X_scaled)

sns.countplot(data=df, x='ELECTROFACIES')
plt.title('ElectroFacies Distribution')
plt.show()
```

```python
fig, axs = plt.subplots(1, 5, figsize=(18, 20), sharey=True) # subplots of 5 different logs
axs[0].plot(df['GR'], df['DEPTH'], color='green')
axs[0].set_title('GR')

axs[1].plot(df['VSH'], df['DEPTH'], color='yellow')
axs[1].set_title('V Shale')

axs[2].plot(df['ILD'], df['DEPTH'], color='blue')
axs[2].set_title('ILD')

axs[3].plot(df['NPHI'], df['DEPTH'], color='red')
axs[3].set_title('NPHI')


axs[4].scatter(df['ELECTROFACIES'], df['DEPTH'], c=df['ELECTROFACIES'], cmap='tab10', s=10)
axs[4].set_title('Electrofacies')
plt.gca().invert_yaxis()  # depth increases downward
plt.show()
```

## 5. Predicting permeability values

### 5a. Calculating permeability ( synthetic permebaility)

```python
# Generate synthetic permeability (realistic Kozeny-Carman style relation + noise)
np.random.seed(42)
df['SYNTH_PERM'] = 10 ** (8 * df['PHI'] - 2.5 + np.random.normal(0, 0.8, len(df)))
df['SYNTH_PERM'] = df['SYNTH_PERM'].clip(0.01, 10000)   # mD range
```

### 5b. ML using XGBoost to predict permeability

```python
# Train XGBoost regressor (logs + facies → permeability)
features = ['GR', 'DPHI', 'NPHI', 'ILD', 'PHI', 'VSH', 'ELECTROFACIES']
X = df[features]
y = np.log10(df['SYNTH_PERM'] + 1)   # log transform for stability

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=29)

#default parameters were used in the regressor arguments
model_perm = xgb.XGBRegressor(n_estimators=100, learning_rate=0.05, max_depth=6, random_state=42)
model_perm.fit(X_train, y_train)

# Predict on full well
df['PRED_PERM'] = 10 ** model_perm.predict(X) - 1
df['PRED_PERM'] = df['PRED_PERM'].clip(0.01, 10000)

print("Permeability prediction RMSE (log scale):", mean_squared_error(y_test, model_perm.predict(X_test), squared=False))
```

### Permeability Prediction using Machine Learning

Permeability was estimated using a machine learning model trained on available petrophysical inputs. This step is critical because permeability governs fluid flow and directly influences reservoir performance.

The model captures nonlinear relationships between porosity and permeability, which are difficult to model using empirical correlations alone. The predicted permeability distribution is subsequently used in uncertainty propagation.

This approach reflects modern reservoir workflows, where data-driven models complement traditional petrophysical interpretations.

## 6. Generate Ensemble of Realizations (Monte Carlo)

```python
# Generate Ensemble of Realizations (Monte Carlo)
n_realizations = 150
metrics_list = []

np.random.seed(42)   # keeps results reproducible

for i in range(n_realizations):
    df_real = df.copy()
    
    # Global systematic bias (affects entire well)
    bias_phi = np.random.normal(0, 0.035)                    # ±2.2 porosity units
    df_real['PHI_noisy'] = df_real['PHI'] + bias_phi
    df_real['PHI_noisy'] = df_real['PHI_noisy'].clip(0, 0.35)
    
    bias_perm_factor = np.random.lognormal(0.0, 0.28)
    df_real['PERM_noisy'] = df_real['PRED_PERM'] * bias_perm_factor
    
    # === CRITICAL: Use the NOISY columns here ===
    avg_phi = df_real['PHI_noisy'].mean()
    ntg = (df_real['PHI_noisy'] > 0.08).mean()
    pore_volume_proxy = df_real['PHI_noisy'].sum() * 0.1524
    perm_geo_mean = np.exp(np.mean(np.log(np.clip(df_real['PERM_noisy'], 1e-6, None))))

   
    # Define Net Pay (based on porosity cutoff)
    net_pay = (df_real['PHI_noisy'] > 0.08).sum() * 0.1524   # thickness in meters

    # Approximate water saturation (since you don't have Sw explicitly)
    # simple assumption: higher porosity → lower Sw
    avg_sw = np.clip(1 - avg_phi, 0.2, 0.8)
    
          

    # Use your already computed permeability
    avg_perm = perm_geo_mean

    # Append metrics
    metrics_list.append({
        'NetPay': net_pay,
        'AvgPhi': avg_phi,
        'Sw': avg_sw,
        'Perm': avg_perm,
        'NTG': ntg,
        'PoreVol_Proxy': pore_volume_proxy,  # keep for reference
        'HCPV': net_pay * avg_phi * (1 - avg_sw),
        'Kh': avg_perm * net_pay
    })

metrics_df = pd.DataFrame(metrics_list)
```

### Interpretation: Monte Carlo-Based Uncertainty Generation

Monte Carlo simulation was used to generate multiple realizations of reservoir properties by introducing stochastic perturbations to porosity and permeability.

Two types of uncertainty were considered:
- Global porosity bias, representing systematic geological uncertainty
- Multiplicative permeability variation, capturing heterogeneity and measurement uncertainty

This approach allows the generation of a suite of plausible reservoir models (150 realizations), representing uncertainty in subsurface properties.

The resulting ensemble forms the basis for uncertainty quantification and model reduction.

### Reservoir Performance Metrics

Key reservoir metrics were computed for each realization to capture both volumetric and flow characteristics:

- Net Pay: Represents effective reservoir thickness contributing to production
- Average Porosity (AvgPhi): Indicates storage capacity
- Hydrocarbon Pore Volume (HCPV): Estimates the volume of hydrocarbons in place
- Flow Capacity (Kh): Reflects the ability of the reservoir to transmit fluids

These metrics move beyond basic petrophysical properties and provide a more physically meaningful representation of reservoir performance potential.

They serve as the foundation for clustering and uncertainty analysis.

### 6b. Simulate the reservoir conditions

```python
# Dynamic proxy
metrics_df['RF_proxy'] = (
    0.25 * (metrics_df['NetPay'] / metrics_df['NetPay'].max()) +
    0.25 * metrics_df['AvgPhi'] +
    0.25 * (np.log(metrics_df['Kh'] + 1) / np.log(np.clip(metrics_df['Kh'], 1e-6, None)) -
    0.25 * metrics_df['Sw']))



metrics_df['NPV_proxy'] = metrics_df['RF_proxy'] * metrics_df['HCPV']
```

### Proxy-Based Reservoir Performance Modeling

In the absence of a full dynamic reservoir simulator, proxy models were introduced to approximate reservoir performance.

A Recovery Factor proxy (RF_proxy) was defined as a function of:
- Net Pay
- Porosity
- Flow capacity (Kh)
- Water saturation (Sw)

The Net Present Value proxy (NPV_proxy) was then computed as:

NPV_proxy = RF_proxy × HCPV

This formulation captures both volumetric and recovery efficiency aspects of the reservoir.

While simplified, this proxy mimics the behavior of dynamic simulation outputs and enables uncertainty analysis in a computationally efficient manner.

## 7. Of the 100 models, select 8 representatives 

```python
## 6. Select Representative Models (Clustering on Static Metrics)
scaler_metrics = StandardScaler()
metrics_scaled = scaler_metrics.fit_transform(metrics_df)   # this is a NumPy array!

kmeans_rep = KMeans(n_clusters=8, random_state=42, n_init=10)
metrics_df['Cluster'] = kmeans_rep.fit_predict(metrics_scaled)

# Select medoid (closest point to cluster center) for each cluster
representatives = []
for cluster in range(8):
    # IMPORTANT: index the scaled NUMPY ARRAY, not the scaler object
    cluster_mask = metrics_df['Cluster'] == cluster
    cluster_points_scaled = metrics_scaled[cluster_mask]          # ← correct: array[boolean_mask]
    
    center = kmeans_rep.cluster_centers_[cluster]
    
    # Find the point in this cluster closest to its own center
    distances = np.linalg.norm(cluster_points_scaled - center, axis=1)
    medoid_local_idx = distances.argmin()                        # index within this cluster
    
    # Convert back to original DataFrame index
    medoid_global_idx = metrics_df[cluster_mask].index[medoid_local_idx]
    
    representatives.append(medoid_global_idx)

print("Selected 8 representative realization indices:", representatives)
```

### Clustering of Reservoir Realizations

KMeans clustering was applied to group reservoir realizations based on their static and derived performance metrics.

This step reduces the complexity of the ensemble by identifying patterns and similarities among realizations. Each cluster represents a subset of models with similar reservoir characteristics.

Clustering is a key step in model reduction workflows, allowing the selection of representative models without losing critical uncertainty information.

## 8. Quantifying uncertainty reduction

```python
full_ensemble_pv = metrics_df['PoreVol_Proxy']
rep_pv = metrics_df.loc[representatives, 'PoreVol_Proxy']

print("FULL ENSEMBLE (100 models):")
print(f"P10 = {np.percentile(full_ensemble_pv, 10):.0f}")
print(f"P50 = {np.percentile(full_ensemble_pv, 50):.0f}")
print(f"P90 = {np.percentile(full_ensemble_pv, 90):.0f}")

print("\nREPRESENTATIVE MODELS (8 models):")
print(f"P10 = {np.percentile(rep_pv, 10):.0f}")
print(f"P50 = {np.percentile(rep_pv, 50):.0f}")
print(f"P90 = {np.percentile(rep_pv, 90):.0f}")

# Plot
plt.figure()
sns.kdeplot(full_ensemble_pv, label='Full Ensemble (100)', shade=True)
sns.kdeplot(rep_pv, label='Representatives (8)', shade=True)
plt.title('Uncertainty Reduction in Pore Volume')
plt.legend()
plt.show()
```

### Uncertainty Quantification (P10, P50, P90)

The P10, P50, and P90 values were computed for both the full ensemble and the representative models using the NPV proxy.

- P10 represents optimistic scenarios (high performance)
- P50 represents the median case
- P90 represents conservative scenarios (low performance)

A close match between the full ensemble and representative models indicates that the reduced model set successfully preserves the uncertainty distribution.

This validates the effectiveness of the clustering and model selection approach.

```python
target = 'NPV_proxy'

full = metrics_df[target]
rep = metrics_df.loc[representatives, target]

print("FULL ENSEMBLE:")
print(f"P10 = {np.percentile(full, 10):.2f}")
print(f"P50 = {np.percentile(full, 50):.2f}")
print(f"P90 = {np.percentile(full, 90):.2f}")

print("\nREPRESENTATIVE MODELS:")
print(f"P10 = {np.percentile(rep, 10):.2f}")
print(f"P50 = {np.percentile(rep, 50):.2f}")
print(f"P90 = {np.percentile(rep, 90):.2f}")

# Plot
plt.figure(figsize=(8,6))
plt.hist(full, bins=20, alpha=0.5, label='Full Ensemble')
plt.hist(rep, bins=8, alpha=0.7, label='Representatives')
plt.legend()
plt.title("NPV Proxy Distribution")
plt.show()
```

### Distribution Comparison

The histogram comparison between the full ensemble and representative models shows how well the reduced set captures the overall uncertainty distribution.

A good match in shape and spread indicates that:
- Extreme cases are preserved
- Central tendencies are maintained
- No significant bias is introduced during model reduction

This visual validation complements the percentile-based comparison.

```python
def compare_spread(full, rep):
    return {
        'Full Range': full.max() - full.min(),
        'Rep Range': rep.max() - rep.min(),
        'Coverage %': (rep.max() - rep.min()) / (full.max() - full.min()) * 100
    }

print(compare_spread(full, rep))
```

### Uncertainty Coverage

The coverage metric quantifies how much of the full ensemble uncertainty range is captured by the representative models.

A high coverage percentage (typically above 70%) indicates that the selected models adequately span the uncertainty space.

This metric provides a quantitative measure of the effectiveness of the model reduction process.
